## 6️⃣ GitHub 토큰 준비

이 실습에서는 GitHub API와 Copilot SDK 인증을 위해 GitHub Personal Access Token이 필요합니다.

권장 권한은 다음과 같습니다.

- Issues: Read and Write
- Contents: Read

매끄러운 실습을 위해서는 다음과 같이 Repository 관련 권한을 모두 할당한 뒤 진행해도 됩니다.    
PAT가 만들어지면 반드시 따로 복사를 해 둡니다. 생성된 토큰은 이후 .env 파일에서 사용해야 합니다.

<img src="images/settings.png" width="800" />

## 7️⃣ .env 파일 설정

프로젝트 루트에 `.env` 파일을 만들고 아래처럼 설정합니다.

```env
GITHUB_TOKEN=ghp_your_token_here
```

혹은 `.env.example` 파일을 활용해서 시작해도 됩니다.

```bash
copy .env.example .env
```

또는 macOS / Linux에서는:

```bash
cp .env.example .env
```

````python
# .env에서 토큰이 정상적으로 읽히는지 확인합니다.
import os
from dotenv import load_dotenv

load_dotenv()
token = os.environ.get('GITHUB_TOKEN') or os.environ.get('GH_TOKEN')
print('GitHub PAT 상태:', '성공 ✅' if token else '미설정 ⚠️')
````

## 8️⃣ Copilot CLI 실행 및 인증 테스트

터미널에서 다음 명령을 실행합니다.

```powershell
copilot
```

- 처음이라면 현재 폴더를 신뢰할지 묻습니다 -> **Trust** 선택
- 프롬프트가 뜨면 로그인:

```text
> /login
```

브라우저가 열리고 터미널에 표시된 일회용 코드(`ABCD-1234` 형식)를 입력 -> **Authorize** -> 터미널로 돌아오면 로그인 완료입니다.

> ✅ 로그인은 한 번만 하면 됩니다. 토큰이 만료되면 다시 `/login`.

**동작 확인**

Copilot 세션 안에서 간단한 프롬프트를 보내 봅니다.

```text
> Say hello and tell me what you can help with
```

응답이 오면 성공입니다. 종료는 아래처럼 합니다.

```text
> /exit
```
